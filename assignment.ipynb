```python
dept_count = df['Department'].value_counts()
dept_count
```

```python

plt.figure(figsize=(8,5))
plt.bar(dept_count.index, dept_count.values)
```

```python

plt.title('Number of Employees in Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
```

```python
 for i, value in enumerate(dept_count.values):
    plt.text(i, value, str(value), ha='center', va='bottom')

plt.show()
```

```python
attrition_count = df['Attrition'].value_counts()
attrition_count
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    attrition_count,
    labels=['No', 'Yes'],
    autopct='%1.1f%%',
    startangle=90
)
```

```python
plt.title('Employee Attrition Distribution')

plt.show()
```

```python
attrition_count = df['Attrition'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    attrition_count.values,
    labels=attrition_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage Distribution of Employee Attrition')
plt.show()
```

```python
plt.figure(figsize=(8,5))

counts, bins, patches = plt.hist(
    df['Age'],
    bins=10,
    edgecolor='black'
)
```

```python
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.title('Age Distribution of Employees')

plt.show()
```

```python
max_index = counts.argmax()

print("Highest Employee Age Range:",
      bins[max_index], "to", bins[max_index + 1])

print("Number of Employees:", int(counts[max_index]))
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Age'],
    df['MonthlyIncome'],
    alpha=0.6
)
```

```python
plt.title('Relationship Between Age and Monthly Income')
plt.xlabel('Age')
plt.ylabel('Monthly Income')

plt.show()
```

```python
avg_income = df.groupby('TotalWorkingYears')['MonthlyIncome'].mean()
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    avg_income.index,
    avg_income.values,
    marker='o',
    linestyle='-'
)
```

```python
plt.title('Average Monthly Income vs Work Experience')
plt.xlabel('Total Working Years')
plt.ylabel('Average Monthly Income')

plt.grid(True)

plt.show()
```
