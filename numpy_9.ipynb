```python
import numpy as np
```

```python
# view vs copy
# view means mdification in original data while copy means
# changs in duplicate data will not reflect original data
```

```python
a = np.array([10,20,30,40,50,60,70,80,90])
a
```

```python
b = a[3:6]
b[:] = 0
```

```python
a
```

```python
# copy
a = np.array([10,20,30,40,50,60,70,80,90])
a
```

```python
b = a[3:6].copy()
b[:] = 0
```

```python
a
```
